# Project 5: Text Similarity & Duplicate Detection — EDA & Model Evaluation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.train_and_evaluate import load_or_fetch_dataset, train_and_evaluate

df = load_or_fetch_dataset()
print(df.info())
df.head()
```

```python
# 1. Class Distribution Plot
plt.figure(figsize=(6, 4))
sns.countplot(x='is_duplicate', data=df, palette='slate')
plt.title('Duplicate vs Non-Duplicate Distribution')
plt.xticks([0, 1], ['Not Duplicate (0)', 'Duplicate (1)'])
plt.show()

# 2. Question Word Length Distribution
df['q1_len'] = df['question1'].astype(str).apply(lambda x: len(x.split()))
df['q2_len'] = df['question2'].astype(str).apply(lambda x: len(x.split()))
plt.figure(figsize=(8, 4))
sns.histplot(df['q1_len'], bins=30, color='steelblue', label='Question 1', alpha=0.6)
sns.histplot(df['q2_len'], bins=30, color='slategray', label='Question 2', alpha=0.6)
plt.xlim(0, 50)
plt.legend()
plt.title('Word Count Distribution')
plt.show()
```
